```python
def analyze_catastrophic_forgetting(df, terrain_name):
    print(f"\n{'='*70}")
    print(f"  CATASTROPHIC FORGETTING ANALYSIS: {terrain_name.upper()} TERRAIN")
    print(f"{'='*70}\n")
    
    # Filter terrain
    df_terr = df[df['Terrain'] == terrain_name].copy()
    
    # Pivot dataframe to pair pre and post by Seed and Arch
    piv = df_terr.pivot_table(
        index=['Seed', 'Arch.'], 
        columns='Stage', 
        values='Total_Reward'
    ).reset_index()
    
    # Compute Forgetting Metrics per seed
    piv['Forgetting_Delta'] = piv['pre'] - piv['post']  # Lower = better
    piv['Retention_Rate'] = (piv['post'] / piv['pre']) * 100  # Higher = better
    
    # Separate architectures across seeds (N=5)
    dac = piv[piv['Arch.'] == 'DAC'].sort_values('Seed')
    uac = piv[piv['Arch.'] == 'UAC'].sort_values('Seed')
    
    # ---------------------------------------------------------
    # 1. Summary Statistics (Mean ± Std)
    # ---------------------------------------------------------
    print("--- Summary Metrics (Across Seeds) ---")
    print(f"DAC Pre-Adaptation  : {dac['pre'].mean():.2f} ± {dac['pre'].std():.2f}")
    print(f"DAC Post-Adaptation : {dac['post'].mean():.2f} ± {dac['post'].std():.2f}")
    print(f"DAC Forgetting (Δ)  : {dac['Forgetting_Delta'].mean():.2f} ± {dac['Forgetting_Delta'].std():.2f}")
    print(f"DAC Retention Rate  : {dac['Retention_Rate'].mean():.1f}% ± {dac['Retention_Rate'].std():.1f}%\n")
    
    print(f"UAC Pre-Adaptation  : {uac['pre'].mean():.2f} ± {uac['pre'].std():.2f}")
    print(f"UAC Post-Adaptation : {uac['post'].mean():.2f} ± {uac['post'].std():.2f}")
    print(f"UAC Forgetting (Δ)  : {uac['Forgetting_Delta'].mean():.2f} ± {uac['Forgetting_Delta'].std():.2f}")
    print(f"UAC Retention Rate  : {uac['Retention_Rate'].mean():.1f}% ± {uac['Retention_Rate'].std():.1f}%\n")
    
    # ---------------------------------------------------------
    # 2. Paired t-tests
    # ---------------------------------------------------------
    print("--- Statistical Tests (Paired t-test, N=5) ---")
    
    # Test A: Does DAC suffer significant forgetting?
    t_dac, p_dac = stats.ttest_rel(dac['pre'], dac['post'])
    print(f"DAC Pre vs Post Drop    : t = {t_dac:.4f}, p = {p_dac:.4f} "
          f"{'(Significant)' if p_dac < 0.05 else '(Not Significant)'}")
    
    # Test B: Does UAC suffer significant forgetting?
    t_uac, p_uac = stats.ttest_rel(uac['pre'], uac['post'])
    print(f"UAC Pre vs Post Drop    : t = {t_uac:.4f}, p = {p_uac:.4f} "
          f"{'(Significant)' if p_uac < 0.05 else '(Not Significant)'}")
    
    # Test C: Is UAC's forgetting delta significantly different from DAC's?
    t_delta, p_delta = stats.ttest_rel(dac['Forgetting_Delta'].to_numpy(), 
                                       uac['Forgetting_Delta'].to_numpy())
    print(f"DAC Δ vs UAC Δ Paired   : t = {t_delta:.4f}, p = {p_delta:.4f} "
          f"{'(Significant Difference)' if p_delta < 0.05 else '(Parity)'}")

# Run analysis for both terrains
analyze_catastrophic_forgetting(df, 'Flat')
analyze_catastrophic_forgetting(df, 'Stair')
```
